```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Đọc ảnh a.jpg
img = cv2.imread('a.jpg')
if img is None:
    print("Không thể đọc file a.jpg. Vui lòng kiểm tra đường dẫn.")
    exit()

# 1. Mean Filter
print("1. Áp dụng Mean Filter...")
mean_filtered = cv2.blur(img, (5, 5))  # Kernel 5x5

# Hiển thị kết quả mean filter
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Ảnh gốc')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(mean_filtered, cv2.COLOR_BGR2RGB))
plt.title('Mean Filter')
plt.axis('off')
plt.show()

# 2. Edge Detection Filter
print("2. Phát hiện biên...")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Sử dụng Canny edge detection
edges_canny = cv2.Canny(gray, 100, 200)

# Sử dụng Sobel filter
sobel_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
sobel_combined = np.sqrt(sobel_x**2 + sobel_y**2)

# Hiển thị kết quả edge detection
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.imshow(gray, cmap='gray')
plt.title('Ảnh xám gốc')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(edges_canny, cmap='gray')
plt.title('Canny Edge Detection')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(sobel_combined, cmap='gray')
plt.title('Sobel Edge Detection')
plt.axis('off')
plt.show()

# 3. Đổi màu ngẫu nhiên
print("3. Đổi màu ngẫu nhiên...")
# Tách các kênh màu BGR
b, g, r = cv2.split(img)

# Trộn ngẫu nhiên các kênh màu
channels = [b, g, r]
np.random.shuffle(channels)

# Tạo ảnh với màu ngẫu nhiên
random_color_img = cv2.merge(channels)

# Lưu ảnh màu ngẫu nhiên
cv2.imwrite('a_random_color.jpg', random_color_img)
print("Đã lưu ảnh màu ngẫu nhiên vào 'a_random_color.jpg'")

# Hiển thị kết quả
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Ảnh gốc')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(random_color_img, cv2.COLOR_BGR2RGB))
plt.title('Ảnh màu ngẫu nhiên')
plt.axis('off')
plt.show()

# 4. Chuyển sang HSV và tách kênh
print("4. Chuyển sang HSV và tách kênh...")
# Chuyển đổi sang không gian màu HSV
hsv_img = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# Tách các kênh HSV
h, s, v = cv2.split(hsv_img)

# Lưu các kênh thành ảnh grayscale
cv2.imwrite('a_hue.jpg', h)
cv2.imwrite('a_saturation.jpg', s)
cv2.imwrite('a_value.jpg', v)

print("Đã lưu các kênh HSV:")
print("- Hue: a_hue.jpg")
print("- Saturation: a_saturation.jpg")
print("- Value: a_value.jpg")

# Hiển thị các kênh HSV
plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Ảnh gốc (BGR)')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(h, cmap='gray')
plt.title('Kênh Hue')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(s, cmap='gray')
plt.title('Kênh Saturation')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(v, cmap='gray')
plt.title('Kênh Value')
plt.axis('off')

plt.tight_layout()
plt.show()

print("Hoàn thành tất cả các yêu cầu!")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import random
import sys

class MultiImageTransformerCLI:
    def __init__(self, image_paths: list):
        """
        Khởi tạo bộ xử lý ảnh.
        
        Args:
            image_paths (list): Danh sách các đường dẫn đến file ảnh.
        """
        self.image_paths = image_paths
        self.original_images = self._load_images()
        
        # Ánh xạ lựa chọn của người dùng tới hàm và tên phương pháp
        self.transformations = {
            'i': ("Inverse", self.inverse_transform),
            'g': ("Gamma Correction", self.gamma_correction),
            'l': ("Log Transformation", self.log_transformation),
            'h': ("Histogram Equalization", self.histogram_equalization),
            'c': ("Contrast Stretching", self.contrast_stretching),
            'a': ("Adaptive Histogram Equalization (CLAHE)", self.adaptive_histogram_equalization)
        }
        # Tên file cho lưu trữ
        self.file_key_map = {
            'i': "inverse",
            'g': "gamma",
            'l': "log",
            'h': "hist_eq",
            'c': "contrast",
            'a': "clahe"
        }

    def _load_images(self) -> list:
        """Tải các ảnh từ đường dẫn và thoát nếu có lỗi."""
        images = []
        print("Đang tải các ảnh đầu vào...")
        for path in self.image_paths:
            img = cv2.imread(path)
            if img is None:
                print(f"--- LỖI ---")
                print(f"Không thể tải ảnh '{path}'.")
                print("Vui lòng đảm bảo các file 'image1.jpg', 'image2.jpg', 'image3.jpg' tồn tại trong cùng thư mục.")
                sys.exit(1) # Thoát chương trình
            images.append(img)
            print(f"  [OK] Đã tải thành công '{path}'")
        
        # Chuẩn hóa kích thước để hiển thị nhất quán
        if images:
            h, w, _ = images[0].shape
            standard_size = (w, h)
            return [cv2.resize(img, standard_size) for img in images]
        return []

    # --- CÁC PHƯƠNG PHÁP BIẾN ĐỔI (Không thay đổi) ---
    def inverse_transform(self, img: np.ndarray) -> np.ndarray:
        return 255 - img
    
    def gamma_correction(self, img: np.ndarray) -> np.ndarray:
        gamma = random.uniform(0.5, 2.0)
        print(f"   - Áp dụng Gamma với gamma={gamma:.2f}")
        normalized = img / 255.0
        corrected = np.power(normalized, gamma)
        return (corrected * 255).astype(np.uint8)
    
    def log_transformation(self, img: np.ndarray) -> np.ndarray:
        c = random.uniform(20.0, 40.0)
        print(f"   - Áp dụng Log với c={c:.2f}")
        log_transformed = c * np.log1p(img.astype(np.float32))
        cv2.normalize(log_transformed, log_transformed, 0, 255, cv2.NORM_MINMAX)
        return log_transformed.astype(np.uint8)
    
    def histogram_equalization(self, img: np.ndarray) -> np.ndarray:
        if len(img.shape) == 3:
            ycrcb = cv2.cvtColor(img, cv2.COLOR_BGR2YCrCb)
            ycrcb[:,:,0] = cv2.equalizeHist(ycrcb[:,:,0])
            return cv2.cvtColor(ycrcb, cv2.COLOR_YCrCb2BGR)
        else:
            return cv2.equalizeHist(img)
    
    def contrast_stretching(self, img: np.ndarray) -> np.ndarray:
        result = np.zeros_like(img, dtype=np.uint8)
        cv2.normalize(img, result, 0, 255, cv2.NORM_MINMAX)
        return result

    def adaptive_histogram_equalization(self, img: np.ndarray) -> np.ndarray:
        clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
        if len(img.shape) == 3:
            ycrcb = cv2.cvtColor(img, cv2.COLOR_BGR2YCrCb)
            ycrcb[:,:,0] = clahe.apply(ycrcb[:,:,0])
            return cv2.cvtColor(ycrcb, cv2.COLOR_YCrCb2BGR)
        else:
            return clahe.apply(img)
            
    def display_results_plt(self, original_images: list, transformed_images: list, method_title: str):
        """Hiển thị kết quả bằng Matplotlib."""
        num_images = len(original_images)
        # Tạo một figure có 2 hàng (gốc và biến đổi) và `num_images` cột
        fig, axes = plt.subplots(2, num_images, figsize=(15, 8))
        fig.suptitle(f"Kết quả biến đổi: {method_title}", fontsize=16)

        for i in range(num_images):
            # Hàng trên: Ảnh gốc
            # Chuyển đổi BGR (OpenCV) sang RGB (Matplotlib)
            original_rgb = cv2.cvtColor(original_images[i], cv2.COLOR_BGR2RGB)
            axes[0, i].imshow(original_rgb)
            axes[0, i].set_title(f"Ảnh gốc {i+1}")
            axes[0, i].axis('off')

            # Hàng dưới: Ảnh đã biến đổi
            transformed_rgb = cv2.cvtColor(transformed_images[i], cv2.COLOR_BGR2RGB)
            axes[1, i].imshow(transformed_rgb)
            axes[1, i].set_title(f"Đã biến đổi {i+1}")
            axes[1, i].axis('off')
        
        plt.tight_layout(rect=[0, 0, 1, 0.96]) # Điều chỉnh để không bị che bởi suptitle
        plt.show()

    def print_menu(self):
        """In menu lựa chọn ra console."""
        print("\n" + "="*50)
        print("CHƯƠNG TRÌNH BIẾN ĐỔI ẢNH")
        print("Nhập ký tự để chọn phép biến đổi:")
        print("  i: Image Inverse Transformation")
        print("  g: Gamma-Correction")
        print("  l: Log Transformation")
        print("  h: Histogram Equalization")
        print("  c: Contrast Stretching")
        print("  a: Adaptive Histogram Equalization")
        print("\n  q: Thoát chương trình")
        print("="*50)

    def run(self):
        """Chạy vòng lặp chính của chương trình."""
        while True:
            self.print_menu()
            choice = input("Nhập lựa chọn của bạn: ").lower().strip()

            if choice == 'q':
                print("Đã thoát chương trình.")
                break

            if choice in self.transformations:
                method_title, method_func = self.transformations[choice]
                file_key = self.file_key_map[choice]
                
                print(f"\n-> Đang áp dụng '{method_title}'...")

                transformed_images = []
                for i, img in enumerate(self.original_images):
                    # Áp dụng hàm biến đổi
                    transformed_img = method_func(img.copy())
                    transformed_images.append(transformed_img)

                    # Lưu ảnh kết quả
                    output_filename = f"output_{file_key}_{i+1}.jpg"
                    cv2.imwrite(output_filename, transformed_img)
                    print(f"   Đã lưu kết quả vào: {output_filename}")
                
                # Hiển thị kết quả bằng Matplotlib
                print("\nHiển thị kết quả... (Hãy đóng cửa sổ ảnh để tiếp tục)")
                self.display_results_plt(self.original_images, transformed_images, method_title)
            else:
                print(f"\n!!! Lựa chọn '{choice}' không hợp lệ. Vui lòng thử lại. !!!")


# --- ĐIỂM BẮT ĐẦU CỦA CHƯƠNG TRÌNH ---
if __name__ == "__main__":
    # 1. Định nghĩa danh sách các ảnh đầu vào
    input_image_files = ['image1.jpg', 'image2.jpg', 'image3.jpg']

    # 2. Khởi tạo và chạy chương trình
    transformer = MultiImageTransformerCLI(input_image_files)
    transformer.run()
```

```python
import cv2
import numpy as np
import os
import matplotlib.pyplot as plt

def display_images_plt(original_img, processed_img, window_title="Processed Image"):
    """
    Hàm hiển thị ảnh gốc và ảnh đã xử lý cạnh nhau bằng Matplotlib.
    Hàm sẽ tự động chuyển đổi màu từ BGR (OpenCV) sang RGB (Matplotlib).
    """
    # LƯU Ý QUAN TRỌNG: OpenCV đọc ảnh là BGR, Matplotlib hiển thị là RGB.
    original_rgb = cv2.cvtColor(original_img, cv2.COLOR_BGR2RGB)
    processed_rgb = cv2.cvtColor(processed_img, cv2.COLOR_BGR2RGB)

    # Tạo một figure với 2 subplot (1 hàng, 2 cột) để hiển thị 2 ảnh
    fig, axes = plt.subplots(1, 2, figsize=(12, 6))
    fig.suptitle(window_title, fontsize=16)

    # Hiển thị ảnh gốc trên subplot bên trái
    axes[0].imshow(original_rgb)
    axes[0].set_title('Ảnh Gốc')
    axes[0].axis('off')  # Ẩn trục tọa độ

    # Hiển thị ảnh đã xử lý trên subplot bên phải
    axes[1].imshow(processed_rgb)
    axes[1].set_title('Ảnh Đã Xử Lý')
    axes[1].axis('off')  # Ẩn trục tọa độ

    plt.tight_layout(rect=[0, 0, 1, 0.96]) # Điều chỉnh layout
    
    # Hiển thị cửa sổ. Chương trình sẽ dừng ở đây cho đến khi bạn đóng cửa sổ này.
    plt.show()

def check_file_exists(filename):
    """Kiểm tra xem tệp có tồn tại không và thông báo lỗi nếu không."""
    if not os.path.exists(filename):
        print(f"Lỗi: Không tìm thấy tệp '{filename}'.")
        print("Vui lòng đảm bảo tệp ảnh có trong cùng thư mục với script.")
        return False
    return True

# --- Bắt đầu xử lý các yêu cầu ---

# 1. Tăng kích thước ảnh colorful-ripe-tropical-fruits.jpg thêm 30 pixel
def task_1_resize_fruits(image_path="colorful-ripe-tropical-fruits.jpg"):
    print("--- Thực hiện Yêu cầu 1 ---")
    if not check_file_exists(image_path):
        return

    # Đọc ảnh gốc
    original_img = cv2.imread(image_path)
    h, w = original_img.shape[:2]
    
    # Kích thước mới
    new_w = w + 30
    new_h = h + 30
    
    # Thay đổi kích thước ảnh
    resized_img = cv2.resize(original_img, (new_w, new_h), interpolation=cv2.INTER_CUBIC)

    # Hiển thị kết quả bằng Matplotlib
    print(f"Hiển thị kết quả cho Yêu cầu 1. Vui lòng đóng cửa sổ ảnh để tiếp tục...")
    display_images_plt(original_img, resized_img, "Yêu cầu 1: Tăng kích thước")

# 2. Xoay ảnh quang-ninh.jpg 45 độ theo chiều kim đồng hồ và lật ngang
def task_2_rotate_flip_quangninh(image_path="quang_ninh.jpg"):
    print("\n--- Thực hiện Yêu cầu 2 ---")
    if not check_file_exists(image_path):
        return
        
    # Đọc ảnh gốc
    original_img = cv2.imread(image_path)
    (h, w) = original_img.shape[:2]
    center = (w // 2, h // 2)

    # Tạo ma trận xoay: góc -45 độ để xoay theo chiều kim đồng hồ
    rotation_matrix = cv2.getRotationMatrix2D(center, -45, 1.0)
    
    # Thực hiện xoay ảnh
    rotated_img = cv2.warpAffine(original_img, rotation_matrix, (w, h))

    # Lật ngang ảnh đã xoay (flipCode = 1)
    flipped_img = cv2.flip(rotated_img, 1)

    # Hiển thị kết quả bằng Matplotlib
    print(f"Hiển thị kết quả cho Yêu cầu 2. Vui lòng đóng cửa sổ ảnh để tiếp tục...")
    display_images_plt(original_img, flipped_img, "Yêu cầu 2: Xoay và Lật ngang")

# 3. Tăng kích thước ảnh pagoda.jpg lên 5 lần và làm mịn bằng Gaussian Blur 7x7
def task_3_upscale_blur_pagoda(image_path="pagoda.jpg"):
    print("\n--- Thực hiện Yêu cầu 3 ---")
    if not check_file_exists(image_path):
        return

    # Đọc ảnh gốc
    original_img = cv2.imread(image_path)
    
    # Tăng kích thước lên 5 lần
    upscaled_img = cv2.resize(original_img, None, fx=5.0, fy=5.0, interpolation=cv2.INTER_CUBIC)
    
    # Áp dụng Gaussian Blur với kernel 7x7
    blurred_img = cv2.GaussianBlur(upscaled_img, (7, 7), 0)
    
    # Để so sánh dễ hơn, thu nhỏ ảnh đã xử lý về kích thước gốc
    h, w = original_img.shape[:2]
    display_blurred = cv2.resize(blurred_img, (w, h))
    
    # Hiển thị kết quả bằng Matplotlib
    print(f"Hiển thị kết quả cho Yêu cầu 3. Vui lòng đóng cửa sổ ảnh để tiếp tục...")
    display_images_plt(original_img, display_blurred, "Yêu cầu 3: Tăng kích thước và Làm mịn")

# 4. Điều chỉnh độ sáng và tương phản cho ảnh pagoda.jpg
def task_4_adjust_contrast_brightness(image_path="pagoda.jpg", alpha=1.5, beta=25):
    print("\n--- Thực hiện Yêu cầu 4 ---")
    if not check_file_exists(image_path):
        return
    
    print(f"Áp dụng công thức: I_out = alpha * I_in + beta với alpha={alpha}, beta={beta}")

    # Đọc ảnh gốc
    original_img = cv2.imread(image_path)
    
    # Áp dụng công thức I_out = a * I_in + b
    adjusted_img = cv2.convertScaleAbs(original_img, alpha=alpha, beta=beta)

    # Hiển thị kết quả bằng Matplotlib
    print(f"Hiển thị kết quả cho Yêu cầu 4. Vui lòng đóng cửa sổ ảnh để tiếp tục...")
    display_images_plt(original_img, adjusted_img, f"Yêu cầu 4: Điều chỉnh Sáng/Tương phản (α={alpha}, β={beta})")


# --- Main execution ---
if __name__ == "__main__":
    task_1_resize_fruits()
    task_2_rotate_flip_quangninh()
    task_3_upscale_blur_pagoda()
    
    # Đối với yêu cầu 4, chọn giá trị ví dụ alpha = 1.5 và beta = 25
    task_4_adjust_contrast_brightness(image_path="pagoda.jpg", alpha=1.5, beta=25)
    
    print("\nĐã hoàn thành tất cả các yêu cầu!")
```
